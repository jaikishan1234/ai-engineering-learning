Create a function "calculate" in which it is having add functionality, subtraction functionality, multiplication and division functionality of two numbers.

```python
def calculate(val1, val2, operation = 'add'):
  if operation == 'add':
    return val1 + val2
  elif operation == 'subtract':
    return val1 - val2
  elif operation == 'multiply':
    return val1 * val2
  elif operation == 'divide':
    return val1 / val2

result = calculate(10, 20, "multiply")
result

'''
def division(a, b):
        if b != 0:
            return a / b
        else:
            return "Division by zero is not allowed"
'''
```

```python
## function definition
def is_armstrong(num):
  str_num = str(num)
  num_digits = len(str_num)

  sum_of_powers = sum(int(digit)**num_digits for digit in str_num)
  return sum_of_powers == num

## function calling
result = is_armstrong(150)
print(result)
```

```python
## function definition
def sum_and_product(num1, num2):
  sum_result = num1 + num2
  product_result = num1 * num2
  return sum_result, product_result

num1 = 6
num2 = 10
sum_result, product_result = sum_and_product(num1, num2)
print(sum_result)
print(product_result)
```

Nested Functions

```python
def calculator(x, y):
  def add(a, b):
    return a + b

  def multiply(a, b):
    return a * b

  def calc(op, a, b):
    if op == 'add':
      return add(a, b)

    else:
      return multiply(a, b)

  return calc('add', x, y), calc('multiply', x, y)

result = calculator(2, 3)
print(result)
```

Task: **kwargs and *args

```python
def print_args(*args):
  for arg in args:
    print(arg)
print_args(10, 20, 30)
```

```python
## function definition
def function_args(**kwargs):
  for key, value in kwargs.items():
    print(f"{key}:{value}")

## function calling
function_args(name='Priya', age=26, salary=30000, city="Bengaluru")
```

Recursive Functions

```python
## function definition
def factorial(num):
  ## base case condition
  if num == 0 or num == 1:
    return 1

  else:
    ## recursive portion
    return num * factorial(num-1)


## function calling
print(factorial(10))
```

Lambda Functions: It is an anonymous function, a concise way of defining small, one-line function.

```python
## lambda arguments : expression
add = lambda x, y : x + y
print(add(2, 3))
```

```python
learner_scores = {
    'Siddharth': 88,
    'Prem': 89,
    'Mohit':93,
    'Swati':90
}

## Sort the above dictionary in the ascending order
sorted_values = dict(sorted(learner_scores.items(), key = lambda item : item[1]))
print(sorted_values)
```

```python
## Sort the above dictionary in the descending order
sorted_values = dict(sorted(learner_scores.items(), key = lambda item : item[1], reverse=True))
print(sorted_values)
```
